```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
total_quantity = df["Quantity"].sum()

profit_margin = (total_profit / total_sales) * 100
average_order_value = total_sales / total_orders
return_rate = (df["Returns"].sum() / len(df)) * 100

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Profit Margin %:", profit_margin)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Quantity:", total_quantity)
print("Average Order Value:", average_order_value)
print("Return Rate %:", return_rate)
```

```python
yearly_performance = (
    df
    .groupby(df["Order Date"].dt.year, as_index=False)
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
    .rename(columns={"Order Date": "Order Year"})
)

yearly_performance["Profit Margin %"] = (
    yearly_performance["Profit"] /
    yearly_performance["Sales"] * 100
)

print(yearly_performance.to_string(index=False))
```

```python
category_performance = (
    df
    .groupby("Category", as_index=False)
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
)

category_performance["Profit Margin %"] = (
    category_performance["Profit"] /
    category_performance["Sales"] * 100
)

print(category_performance.to_string(index=False))
```

```python
subcategory_performance = (
    df
    .groupby("Sub-Category", as_index=False)
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
)

subcategory_performance["Profit Margin %"] = (
    subcategory_performance["Profit"] /
    subcategory_performance["Sales"] * 100
)

print(subcategory_performance.sort_values(
    "Profit Margin %",
    ascending=True
).to_string(index=False))
```

```python
product_performance = (
    df
    .groupby(["Product ID", "Product Name"], as_index=False)
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
)

product_performance["Profit Margin %"] = (
    product_performance["Profit"] /
    product_performance["Sales"] * 100
)

print(
    product_performance
    .sort_values("Profit")
    .head(10)
    .to_string(index=False)
)
```

```python
print(
    product_performance
    .sort_values("Profit", ascending=False)
    .head(10)
    .to_string(index=False)
)
```

```python
customer_performance = (
    df
    .groupby(["Customer ID", "Customer Name"], as_index=False)
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
)

customer_performance["Profit Margin %"] = (
    customer_performance["Profit"] /
    customer_performance["Sales"] * 100
)

print("Top 10 Customers by Sales")
print(
    customer_performance
    .sort_values("Sales", ascending=False)
    .head(10)
    .to_string(index=False)
)

print("\nTop 10 Customers by Profit")
print(
    customer_performance
    .sort_values("Profit", ascending=False)
    .head(10)
    .to_string(index=False)
)
```

```python
region_performance = (
    df
    .groupby("Region", as_index=False)
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
)

region_performance["Profit Margin %"] = (
    region_performance["Profit"] /
    region_performance["Sales"] * 100
)

print(region_performance.to_string(index=False))
```

```python
return_by_category = (
    df
    .groupby("Category", as_index=False)
    .agg(
        Total_Rows=("Returns", "size"),
        Returned_Rows=("Returns", "sum")
    )
)

return_by_category["Return Rate %"] = (
    return_by_category["Returned_Rows"] /
    return_by_category["Total_Rows"] * 100
)

print(return_by_category.to_string(index=False))
```

```python
payment_performance = (
    df
    .groupby("Payment Mode", as_index=False)
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique"),
        Quantity=("Quantity", "sum")
    )
)

payment_performance["Profit Margin %"] = (
    payment_performance["Profit"] /
    payment_performance["Sales"] * 100
)

print(payment_performance.to_string(index=False))
```
